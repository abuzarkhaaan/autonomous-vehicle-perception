# Qualitative Results

Load a prediction JSON (from scripts/inference.py) and create a stub BEV artifact.

```python
from utils.visualization import load_json, save_json, make_stub_bev_artifact

pred_path = '../outputs/predictions'
print('Set pred_path to a specific file like ../outputs/predictions/pred_YYYYMMDD-HHMMSS.json')
```

```python
import glob
from pathlib import Path

cands = sorted(glob.glob('../outputs/predictions/pred_*.json'))
print('found:', len(cands))
if cands:
    p = cands[-1]
    obj = load_json(p)
    artifact = make_stub_bev_artifact(obj.get('predictions', []), meta={'source': p})
    out = Path('../outputs/visualizations') / (Path(p).stem + '_artifact.json')
    save_json(out, artifact)
    print('saved:', out)
```
