# Dataset Exploration

Quick sanity checks for multi-camera inputs and basic dataset stats.

```python
from datasets.nuscenes_dataset import NuScenesDataset

ds = NuScenesDataset(root='../data/nuscenes', split='train')
print('len:', len(ds))
s = ds[0]
print('keys:', list(s.keys()))
print('num_cams:', len(s['images']))
print('image shape:', tuple(s['images'][0].shape))
print('gt_boxes_3d:', s['gt_boxes_3d'].shape)
print('gt_labels:', s['gt_labels'].shape)
```

```python
counts = {}
for i in range(min(len(ds), 50)):
    si = ds[i]
    n = int(si['gt_labels'].numel())
    counts[n] = counts.get(n, 0) + 1
print('box count histogram (first 50):', dict(sorted(counts.items())))
```
